```python
from pathlib import Path
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from tqdm.auto import tqdm
import rasterio

# ====================== PATHS ======================
DATA_ROOT = Path("/home/23CSU413/Deep Learning/AgriFieldNet")

SOURCE_DIR       = DATA_ROOT / "source"
TRAIN_LABEL_DIR  = DATA_ROOT / "train_labels" / "train_labels"   # nested!
TEST_LABEL_DIR   = DATA_ROOT / "test_labels"

print("="*60)
print("PATH VERIFICATION")
print("="*60)
print("DATA_ROOT exists      :", DATA_ROOT.exists())
print("SOURCE_DIR exists     :", SOURCE_DIR.exists())
print("TRAIN_LABEL_DIR exists:", TRAIN_LABEL_DIR.exists())
print("TEST_LABEL_DIR exists :", TEST_LABEL_DIR.exists())
print()

# Count files
n_source = len([p for p in SOURCE_DIR.iterdir() if p.is_dir()])
n_train  = len(list(TRAIN_LABEL_DIR.glob("*.tif")))
n_test   = len(list(TEST_LABEL_DIR.glob("*.tif")))

print(f"Number of source chips      : {n_source}")
print(f"Number of train label files : {n_train}")
print(f"Number of test label files  : {n_test}")
print()

# Show a few train label filenames
print("Sample train label files:")
for f in sorted(TRAIN_LABEL_DIR.glob("*.tif"))[:6]:
    print("  ", f.name)
```

```python
# ====================== BANDS & CROP MAP ======================
BANDS = ["B01", "B02", "B03", "B04", "B05", "B06",
         "B07", "B08", "B8A", "B09", "B11", "B12"]

CROP_MAP = {
    0: "NoData",
    1: "Wheat", 2: "Mustard", 3: "Lentil", 4: "Fallow",
    5: "GreenPea", 6: "Sugarcane", 8: "Garlic", 9: "Maize",
    13: "Gram", 14: "Coriander", 15: "Potato", 16: "Berseem", 36: "Rice"
}

# ====================== GET ALL CHIP IDs ======================
def get_all_chip_ids():
    folders = [p.name for p in SOURCE_DIR.iterdir() if p.is_dir()]
    ids = sorted([f.split("_")[-1] for f in folders])
    return ids

all_chip_ids = get_all_chip_ids()
print(f"Total unique chip IDs from source: {len(all_chip_ids)}")
print("First 10:", all_chip_ids[:10])
print()

# ====================== GET TRAIN CHIP IDs ======================
def get_train_chip_ids():
    files = list(TRAIN_LABEL_DIR.glob("*_field_ids.tif"))
    ids = sorted(set([f.name.split("_")[-3] if "field_ids" in f.name else f.name.split("_")[-1].replace(".tif","") 
                      for f in files]))
    # cleaner way
    ids = []
    for f in files:
        # filename pattern: ..._train_XXXXX_field_ids.tif
        parts = f.stem.split("_")
        ids.append(parts[-3])   # the chip id is before "field_ids"
    return sorted(set(ids))

train_chip_ids = get_train_chip_ids()
print(f"Train chips that have labels: {len(train_chip_ids)}")
print("First 10 train chip IDs:", train_chip_ids[:10])
print()

# ====================== LOADING FUNCTIONS ======================
def load_chip_bands(chip_id):
    folder = SOURCE_DIR / f"ref_agrifieldnet_competition_v1_source_{chip_id}"
    data = {}
    for b in BANDS:
        tif = folder / f"ref_agrifieldnet_competition_v1_source_{chip_id}_{b}_10m.tif"
        with rasterio.open(tif) as src:
            data[b] = src.read(1).astype(np.float32)
    return data

def load_train_labels(chip_id):
    crop_path  = TRAIN_LABEL_DIR / f"ref_agrifieldnet_competition_v1_labels_train_{chip_id}.tif"
    field_path = TRAIN_LABEL_DIR / f"ref_agrifieldnet_competition_v1_labels_train_{chip_id}_field_ids.tif"
    
    with rasterio.open(crop_path) as src:
        crop = src.read(1)
    with rasterio.open(field_path) as src:
        field = src.read(1)
    return crop, field

def load_test_field_ids(chip_id):
    field_path = TEST_LABEL_DIR / f"ref_agrifieldnet_competition_v1_labels_test_{chip_id}_field_ids.tif"
    with rasterio.open(field_path) as src:
        return src.read(1)

# ====================== QUICK SANITY CHECK ======================
print("="*60)
print("SANITY CHECK ON FIRST TRAIN CHIP")
print("="*60)

test_cid = train_chip_ids[0]
print("Testing chip:", test_cid)

bands = load_chip_bands(test_cid)
print("Bands loaded successfully. Shape of B04:", bands["B04"].shape)

crop, field = load_train_labels(test_cid)
print("Crop label unique values:", np.unique(crop))
print("Field ID unique values (first 15):", np.unique(field)[:15])
print("Number of non-zero field pixels:", np.sum(field > 0))
```

```python
print("="*60)
print("STEP 3: EXPLORATORY DATA ANALYSIS")
print("="*60)

# ---------- 3.1 Class distribution at PIXEL level ----------
print("\n[3.1] Computing pixel-level class distribution (this may take 1-2 min)...")

pixel_counts = Counter()
field_pixel_counts = Counter()   # how many pixels per field (for later)

for cid in tqdm(train_chip_ids, desc="Scanning train chips"):
    try:
        crop, field = load_train_labels(cid)
        unique, counts = np.unique(crop, return_counts=True)
        pixel_counts.update(dict(zip(unique, counts)))
        
        # also count non-zero field pixels
        field_pixel_counts.update(field[field > 0])
    except Exception as e:
        print(f"Error on {cid}: {e}")

print("\nPixel-level class counts:")
for cls in sorted(pixel_counts.keys()):
    name = CROP_MAP.get(cls, f"Unknown-{cls}")
    print(f"  {cls:2d} ({name:12s}): {pixel_counts[cls]:>10,}")

# ---------- 3.2 Number of unique fields ----------
print(f"\nTotal unique field IDs across all train chips: {len(field_pixel_counts):,}")

# ---------- 3.3 Visualise class distribution ----------
plt.figure(figsize=(12, 5))
classes = sorted([c for c in pixel_counts.keys() if c != 0])
names   = [CROP_MAP.get(c, str(c)) for c in classes]
values  = [pixel_counts[c] for c in classes]

sns.barplot(x=names, y=values, palette="viridis")
plt.xticks(rotation=45, ha="right")
plt.title("Pixel-level Class Distribution (Train set, excluding NoData)")
plt.ylabel("Number of pixels")
plt.tight_layout()
plt.show()

# ---------- 3.4 Quick look at one chip ----------
print("\n[3.4] Visualising first train chip...")
cid = train_chip_ids[0]
bands = load_chip_bands(cid)
crop, field = load_train_labels(cid)

# True colour RGB (B04, B03, B02) - scale for display
rgb = np.stack([bands["B04"], bands["B03"], bands["B02"]], axis=-1)
rgb = np.clip(rgb / 3000.0, 0, 1)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(rgb)
axes[0].set_title(f"RGB - Chip {cid}")
axes[0].axis("off")

im1 = axes[1].imshow(crop, cmap="tab20", vmin=0, vmax=40)
axes[1].set_title("Crop Labels")
axes[1].axis("off")
plt.colorbar(im1, ax=axes[1], fraction=0.046)

im2 = axes[2].imshow(field, cmap="nipy_spectral")
axes[2].set_title("Field IDs")
axes[2].axis("off")
plt.colorbar(im2, ax=axes[2], fraction=0.046)

plt.suptitle(f"Chip {cid} | Unique crops: {np.unique(crop)}")
plt.tight_layout()
plt.show()

print("\nEDA Step 3 completed successfully.")
```

```python
print("="*60)
print("STEP 4: FIELD-LEVEL FEATURE ENGINEERING")
print("="*60)

def extract_field_features(chip_ids):
    records = []
    
    for cid in tqdm(chip_ids, desc="Extracting field features"):
        try:
            bands = load_chip_bands(cid)
            crop, field = load_train_labels(cid)
            
            # Stack bands → shape (12, 256, 256)
            stack = np.stack([bands[b] for b in BANDS], axis=0)
            
            unique_fids = np.unique(field)
            unique_fids = unique_fids[unique_fids > 0]  # remove background
            
            for fid in unique_fids:
                mask = (field == fid)
                n_pix = int(mask.sum())
                if n_pix < 5:          # skip extremely small fields
                    continue
                
                # Majority crop label inside this field
                crop_vals = crop[mask]
                majority_crop = Counter(crop_vals).most_common(1)[0][0]
                if majority_crop == 0:  # skip if majority is NoData
                    continue
                
                feats = {
                    "field_id": int(fid),
                    "chip_id": cid,
                    "n_pixels": n_pix,
                    "crop_id": int(majority_crop)
                }
                
                # Mean & Std of each band
                for i, b in enumerate(BANDS):
                    vals = stack[i][mask]
                    feats[f"mean_{b}"] = float(vals.mean())
                    feats[f"std_{b}"]  = float(vals.std())
                
                # Vegetation Indices (using mean reflectance)
                nir   = feats["mean_B08"]
                red   = feats["mean_B04"]
                green = feats["mean_B03"]
                blue  = feats["mean_B02"]
                re    = feats["mean_B05"]   # Red-edge
                swir  = feats["mean_B11"]
                
                feats["NDVI"]  = (nir - red) / (nir + red + 1e-6)
                feats["NDWI"]  = (green - nir) / (green + nir + 1e-6)
                feats["EVI"]   = 2.5 * (nir - red) / (nir + 6*red - 7.5*blue + 1 + 1e-6)
                feats["SAVI"]  = 1.5 * (nir - red) / (nir + red + 0.5 + 1e-6)
                feats["GCI"]   = (nir / (green + 1e-6)) - 1
                feats["NDRE"]  = (nir - re) / (nir + re + 1e-6)
                
                records.append(feats)
                
        except Exception as e:
            print(f"Error on chip {cid}: {e}")
            continue
    
    return pd.DataFrame(records)

# Run extraction
train_df = extract_field_features(train_chip_ids)

print("\n" + "="*60)
print("FEATURE EXTRACTION COMPLETED")
print("="*60)
print(f"Total fields extracted : {len(train_df):,}")
print(f"Columns                : {list(train_df.columns)}")
print()
print("Class distribution at FIELD level:")
print(train_df["crop_id"].value_counts().sort_index())
print()
print("First 5 rows:")
print(train_df.head())
```

```python
print("="*60)
print("STEP 5: TRAIN/VAL SPLIT + RANDOM FOREST BASELINE")
print("="*60)

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (classification_report, confusion_matrix,
                             accuracy_score, balanced_accuracy_score, f1_score)
import joblib

# ---------- Prepare data ----------
# Keep only known crop classes
train_df = train_df[train_df["crop_id"].isin(CROP_MAP.keys())]
train_df = train_df[train_df["crop_id"] != 0].copy()

# Encode labels to 0, 1, 2, ...
le = LabelEncoder()
train_df["label"] = le.fit_transform(train_df["crop_id"])

print("Label mapping (crop_id → encoded):")
for orig, enc in zip(le.classes_, le.transform(le.classes_)):
    print(f"  {orig:2d} ({CROP_MAP[orig]:12s}) → {enc}")

# Feature columns
feature_cols = [c for c in train_df.columns if c.startswith(("mean_", "std_", "NDVI", "NDWI", "EVI", "SAVI", "GCI", "NDRE"))]
print(f"\nNumber of features: {len(feature_cols)}")

X = train_df[feature_cols].values
y = train_df["label"].values

# Stratified split
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print(f"\nTrain samples: {len(X_train):,}")
print(f"Val samples  : {len(X_val):,}")

# Scale features (important for SVM later, also good practice)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled   = scaler.transform(X_val)
```

```python
# ---------- Random Forest ----------
print("\nTraining Random Forest...")
rf = RandomForestClassifier(
    n_estimators=400,
    max_depth=22,
    min_samples_leaf=3,
    class_weight="balanced_subsample",
    n_jobs=-1,
    random_state=42,
    verbose=1
)
rf.fit(X_train_scaled, y_train)

# Predict
y_pred = rf.predict(X_val_scaled)

# Metrics
print("\n" + "="*60)
print("RANDOM FOREST RESULTS")
print("="*60)
print(f"Accuracy          : {accuracy_score(y_val, y_pred):.4f}")
print(f"Balanced Accuracy : {balanced_accuracy_score(y_val, y_pred):.4f}")
print(f"Macro F1          : {f1_score(y_val, y_pred, average='macro'):.4f}")
print(f"Weighted F1       : {f1_score(y_val, y_pred, average='weighted'):.4f}")

print("\nClassification Report:")
print(classification_report(
    y_val, y_pred,
    target_names=[CROP_MAP[c] for c in le.classes_]
))

# Save model + scaler + label encoder
joblib.dump(rf, "rf_baseline.joblib")
joblib.dump(scaler, "scaler.joblib")
joblib.dump(le, "label_encoder.joblib")
print("\nModels saved: rf_baseline.joblib, scaler.joblib, label_encoder.joblib")
```

```python
print("="*60)
print("STEP 6: SVM BASELINE")
print("="*60)

from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, balanced_accuracy_score, f1_score
import time

# We already have X_train_scaled, X_val_scaled, y_train, y_val from previous step

print("Training SVM (this may take 1–3 minutes)...")
start = time.time()

svm = SVC(
    kernel="rbf",
    C=10,
    gamma="scale",
    class_weight="balanced",
    probability=False,          # set True only if you need probabilities later
    random_state=42,
    verbose=True
)

svm.fit(X_train_scaled, y_train)

print(f"Training time: {time.time() - start:.1f} seconds")

# Predict
y_pred_svm = svm.predict(X_val_scaled)
```

```python
print("="*60)
print("STEP 7: ARTIFICIAL NEURAL NETWORK (MLP)")
print("="*60)

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# We already have:
# X_train_scaled, X_val_scaled, y_train, y_val, le

n_features = X_train_scaled.shape[1]
n_classes  = len(le.classes_)

print(f"Input features : {n_features}")
print(f"Number of classes: {n_classes}")

# Class weights for imbalance
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = dict(enumerate(class_weights))
print("\nClass weights:", {le.classes_[k]: round(v,2) for k,v in class_weight_dict.items()})

# Build MLP
def build_mlp(input_dim, n_classes):
    model = models.Sequential([
        layers.Input(shape=(input_dim,)),
        layers.Dense(256, activation="relu"),
        layers.BatchNormalization(),
        layers.Dropout(0.35),
        
        layers.Dense(128, activation="relu"),
        layers.BatchNormalization(),
        layers.Dropout(0.30),
        
        layers.Dense(64, activation="relu"),
        layers.BatchNormalization(),
        layers.Dropout(0.25),
        
        layers.Dense(n_classes, activation="softmax")
    ])
    
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

mlp = build_mlp(n_features, n_classes)
mlp.summary()

# Callbacks
early_stop = callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=8,
    min_lr=1e-6,
    verbose=1
)

# Train
print("\nTraining MLP...")
history = mlp.fit(
    X_train_scaled, y_train,
    validation_data=(X_val_scaled, y_val),
    epochs=150,
    batch_size=128,
    class_weight=class_weight_dict,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# Evaluate
y_pred_proba = mlp.predict(X_val_scaled)
y_pred_mlp = np.argmax(y_pred_proba, axis=1)
```

```python
print("\n" + "="*60)
print("MLP / ANN RESULTS")
print("="*60)
print(f"Accuracy          : {accuracy_score(y_val, y_pred_mlp):.4f}")
print(f"Balanced Accuracy : {balanced_accuracy_score(y_val, y_pred_mlp):.4f}")
print(f"Macro F1          : {f1_score(y_val, y_pred_mlp, average='macro'):.4f}")
print(f"Weighted F1       : {f1_score(y_val, y_pred_mlp, average='weighted'):.4f}")

print("\nClassification Report:")
print(classification_report(
    y_val, y_pred_mlp,
    target_names=[CROP_MAP[c] for c in le.classes_],
    zero_division=0
))

# Save model
mlp.save("mlp_crop_classifier.keras")
print("\nMLP model saved: mlp_crop_classifier.keras")
```

```python
print("="*60)
print("STEP 8A: PREPARING PATCHES FOR CNN")
print("="*60)

import numpy as np
from tqdm.auto import tqdm

PATCH_SIZE = 32          # 32x32 pixels around field centroid
HALF = PATCH_SIZE // 2

def get_field_centroid(field_mask):
    """Return (row, col) centroid of a binary mask"""
    rows, cols = np.where(field_mask)
    if len(rows) == 0:
        return None
    return int(rows.mean()), int(cols.mean())

def extract_patch(stack, centroid, patch_size=PATCH_SIZE):
    """Extract a patch centered on centroid, with padding if needed"""
    r, c = centroid
    h, w = stack.shape[1], stack.shape[2]
    
    r1 = max(0, r - HALF)
    r2 = min(h, r + HALF)
    c1 = max(0, c - HALF)
    c2 = min(w, c + HALF)
    
    patch = stack[:, r1:r2, c1:c2]
    
    # Pad if the patch is smaller than PATCH_SIZE (near borders)
    pad_r = PATCH_SIZE - patch.shape[1]
    pad_c = PATCH_SIZE - patch.shape[2]
    
    if pad_r > 0 or pad_c > 0:
        patch = np.pad(patch, 
                       ((0,0), (0, max(0,pad_r)), (0, max(0,pad_c))), 
                       mode='reflect')
    
    return patch[:, :PATCH_SIZE, :PATCH_SIZE]

# We will extract patches only for a manageable subset first (or all if memory allows)
# For full dataset it can take 5-15 minutes and several GB RAM

print("Extracting 32x32 patches for all fields (this may take several minutes)...")

patches = []
patch_labels = []
patch_field_ids = []

for cid in tqdm(train_chip_ids, desc="Chips"):
    try:
        bands = load_chip_bands(cid)
        crop, field = load_train_labels(cid)
        
        stack = np.stack([bands[b] for b in BANDS], axis=0)  # (12, 256, 256)
        
        unique_fids = np.unique(field)
        unique_fids = unique_fids[unique_fids > 0]
        
        for fid in unique_fids:
            mask = (field == fid)
            if mask.sum() < 5:
                continue
            
            # majority crop
            majority = Counter(crop[mask]).most_common(1)[0][0]
            if majority == 0:
                continue
            
            centroid = get_field_centroid(mask)
            if centroid is None:
                continue
            
            patch = extract_patch(stack, centroid)
            
            # Simple normalisation per patch (important for CNN)
            patch = patch / 10000.0          # Sentinel-2 reflectance scaling
            patch = np.clip(patch, 0, 1)
            
            patches.append(patch)
            patch_labels.append(majority)
            patch_field_ids.append(fid)
            
    except Exception as e:
        print(f"Error on {cid}: {e}")
        continue

patches = np.array(patches)                # (N, 12, 32, 32)
patch_labels = np.array(patch_labels)

print(f"\nExtracted patches shape: {patches.shape}")
print(f"Labels shape          : {patch_labels.shape}")
print("Unique crops in patches:", np.unique(patch_labels))
```

```python
print("="*60)
print("STEP 8B: CNN TRAINING")
print("="*60)

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score, balanced_accuracy_score, f1_score
import numpy as np

# ---------- Prepare data ----------
# Encode labels the same way as before
le_cnn = LabelEncoder()
y_cnn = le_cnn.fit_transform(patch_labels)

print("Label mapping for CNN:")
for orig, enc in zip(le_cnn.classes_, le_cnn.transform(le_cnn.classes_)):
    print(f"  {orig:2d} ({CROP_MAP[orig]:12s}) → {enc}")

# Train / Val split (stratified)
X_tr, X_val, y_tr, y_val = train_test_split(
    patches, y_cnn, test_size=0.20, stratify=y_cnn, random_state=42
)

# Keras expects channels-last → (N, 32, 32, 12)
X_tr  = np.transpose(X_tr,  (0, 2, 3, 1))
X_val = np.transpose(X_val, (0, 2, 3, 1))

print(f"\nTrain patches: {X_tr.shape}")
print(f"Val patches  : {X_val.shape}")

# Class weights
cw = compute_class_weight("balanced", classes=np.unique(y_tr), y=y_tr)
class_weight_dict = dict(enumerate(cw))

# ---------- Build a simple but effective CNN ----------
def build_cnn(input_shape=(32, 32, 12), n_classes=13):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        
        # Block 1
        layers.Conv2D(32, (3,3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.Conv2D(32, (3,3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Dropout(0.25),
        
        # Block 2
        layers.Conv2D(64, (3,3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.Conv2D(64, (3,3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2,2)),
        layers.Dropout(0.30),
        
        # Block 3
        layers.Conv2D(128, (3,3), padding="same", activation="relu"),
        layers.BatchNormalization(),
        layers.GlobalAveragePooling2D(),
        layers.Dropout(0.40),
        
        # Classifier
        layers.Dense(128, activation="relu"),
        layers.BatchNormalization(),
        layers.Dropout(0.30),
        layers.Dense(n_classes, activation="softmax")
    ])
    
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-3),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )
    return model

cnn = build_cnn(input_shape=(32, 32, 12), n_classes=len(le_cnn.classes_))
cnn.summary()

# Callbacks
early_stop = callbacks.EarlyStopping(monitor="val_loss", patience=15, restore_best_weights=True, verbose=1)
reduce_lr  = callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=6, min_lr=1e-6, verbose=1)

# Train
print("\nTraining CNN...")
history = cnn.fit(
    X_tr, y_tr,
    validation_data=(X_val, y_val),
    epochs=80,
    batch_size=64,
    class_weight=class_weight_dict,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# Evaluate
y_pred_proba = cnn.predict(X_val)
y_pred_cnn = np.argmax(y_pred_proba, axis=1)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, accuracy_score, balanced_accuracy_score, f1_score
import numpy as np
import pandas as pd

# ---------- Data (already extracted earlier) ----------
# patches shape: (5573, 12, 32, 32)
# patch_labels shape: (5573,)

le_cnn = LabelEncoder()
y_cnn = le_cnn.fit_transform(patch_labels)

X = np.transpose(patches, (0, 2, 3, 1))   # → (N, 32, 32, 12)
y = y_cnn

X_tr, X_val, y_tr, y_val = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print("Train:", X_tr.shape, "Val:", X_val.shape)

# Class weights
cw = compute_class_weight("balanced", classes=np.unique(y_tr), y=y_tr)
class_weight_dict = dict(enumerate(cw))

n_classes = len(le_cnn.classes_)
INPUT_SHAPE = (32, 32, 12)
```

```python
def build_lenet(input_shape=INPUT_SHAPE, n_classes=n_classes):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        
        layers.Conv2D(6, (5,5), activation="tanh", padding="same"),
        layers.AveragePooling2D((2,2)),
        
        layers.Conv2D(16, (5,5), activation="tanh", padding="same"),
        layers.AveragePooling2D((2,2)),
        
        layers.Flatten(),
        layers.Dense(120, activation="tanh"),
        layers.Dense(84, activation="tanh"),
        layers.Dense(n_classes, activation="softmax")
    ])
    
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def build_alexnet(input_shape=INPUT_SHAPE, n_classes=n_classes):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        
        layers.Conv2D(64, (5,5), strides=1, activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        layers.Conv2D(128, (3,3), activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        layers.Conv2D(256, (3,3), activation="relu", padding="same"),
        layers.Conv2D(256, (3,3), activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        layers.Flatten(),
        layers.Dense(512, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(n_classes, activation="softmax")
    ])
    
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def build_vgg(input_shape=INPUT_SHAPE, n_classes=n_classes):
    model = models.Sequential([
        layers.Input(shape=input_shape),
        
        # Block 1
        layers.Conv2D(64, (3,3), activation="relu", padding="same"),
        layers.Conv2D(64, (3,3), activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        # Block 2
        layers.Conv2D(128, (3,3), activation="relu", padding="same"),
        layers.Conv2D(128, (3,3), activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        # Block 3
        layers.Conv2D(256, (3,3), activation="relu", padding="same"),
        layers.Conv2D(256, (3,3), activation="relu", padding="same"),
        layers.MaxPooling2D((2,2)),
        layers.BatchNormalization(),
        
        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation="relu"),
        layers.Dropout(0.5),
        layers.Dense(n_classes, activation="softmax")
    ])
    
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def inception_module(x, filters):
    f1, f3r, f3, f5r, f5, fp = filters
    
    conv1 = layers.Conv2D(f1, (1,1), padding="same", activation="relu")(x)
    
    conv3 = layers.Conv2D(f3r, (1,1), padding="same", activation="relu")(x)
    conv3 = layers.Conv2D(f3, (3,3), padding="same", activation="relu")(conv3)
    
    conv5 = layers.Conv2D(f5r, (1,1), padding="same", activation="relu")(x)
    conv5 = layers.Conv2D(f5, (5,5), padding="same", activation="relu")(conv5)
    
    pool = layers.MaxPooling2D((3,3), strides=1, padding="same")(x)
    pool = layers.Conv2D(fp, (1,1), padding="same", activation="relu")(pool)
    
    return layers.Concatenate()([conv1, conv3, conv5, pool])

def build_googlenet(input_shape=INPUT_SHAPE, n_classes=n_classes):
    inputs = layers.Input(shape=input_shape)
    
    x = layers.Conv2D(64, (3,3), padding="same", activation="relu")(inputs)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.BatchNormalization()(x)
    
    x = inception_module(x, [32, 32, 64, 16, 32, 32])
    x = inception_module(x, [64, 48, 96, 24, 48, 48])
    x = layers.MaxPooling2D((2,2))(x)
    
    x = inception_module(x, [64, 64, 128, 32, 64, 64])
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(n_classes, activation="softmax")(x)
    
    model = models.Model(inputs, outputs)
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def residual_block(x, filters, stride=1):
    shortcut = x
    x = layers.Conv2D(filters, (3,3), strides=stride, padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    
    x = layers.Conv2D(filters, (3,3), padding="same")(x)
    x = layers.BatchNormalization()(x)
    
    if stride != 1 or shortcut.shape[-1] != filters:
        shortcut = layers.Conv2D(filters, (1,1), strides=stride, padding="same")(shortcut)
        shortcut = layers.BatchNormalization()(shortcut)
    
    x = layers.Add()([x, shortcut])
    x = layers.ReLU()(x)
    return x

def build_resnet(input_shape=INPUT_SHAPE, n_classes=n_classes):
    inputs = layers.Input(shape=input_shape)
    
    x = layers.Conv2D(64, (3,3), padding="same")(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    
    x = residual_block(x, 64)
    x = residual_block(x, 64)
    
    x = residual_block(x, 128, stride=2)
    x = residual_block(x, 128)
    
    x = residual_block(x, 256, stride=2)
    x = residual_block(x, 256)
    
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    outputs = layers.Dense(n_classes, activation="softmax")(x)
    
    model = models.Model(inputs, outputs)
    model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model
```

```python
def train_and_evaluate(model, name, epochs=60, batch_size=64):
    print(f"\n{'='*60}")
    print(f"Training {name}")
    print(f"{'='*60}")
    
    early_stop = callbacks.EarlyStopping(monitor="val_loss", patience=12, restore_best_weights=True, verbose=1)
    reduce_lr  = callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6, verbose=1)
    
    history = model.fit(
        X_tr, y_tr,
        validation_data=(X_val, y_val),
        epochs=epochs,
        batch_size=batch_size,
        class_weight=class_weight_dict,
        callbacks=[early_stop, reduce_lr],
        verbose=1
    )
    
    y_pred = np.argmax(model.predict(X_val), axis=1)
    
    acc  = accuracy_score(y_val, y_pred)
    bacc = balanced_accuracy_score(y_val, y_pred)
    mf1  = f1_score(y_val, y_pred, average="macro")
    wf1  = f1_score(y_val, y_pred, average="weighted")
    
    print(f"\n{name} Results:")
    print(f"Accuracy          : {acc:.4f}")
    print(f"Balanced Accuracy : {bacc:.4f}")
    print(f"Macro F1          : {mf1:.4f}")
    print(f"Weighted F1       : {wf1:.4f}")
    
    print(classification_report(y_val, y_pred,
          target_names=[CROP_MAP[c] for c in le_cnn.classes_], zero_division=0))
    
    model.save(f"{name.lower().replace(' ', '_')}_agrifieldnet.keras")
    return {"Model": name, "Accuracy": acc, "Balanced Accuracy": bacc, "Macro F1": mf1, "Weighted F1": wf1}
```

```python
# ============================================================
# 5-Fold CV + Manual Oversampling (no imblearn needed)
# ============================================================

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import RobustScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
import numpy as np
import pandas as pd

def manual_oversample(X, y, random_state=42):
    """
    Simple random oversampling of minority classes
    so that every class has the same number of samples
    as the majority class.
    """
    np.random.seed(random_state)
    
    df = pd.DataFrame(X)
    df['label'] = y
    
    # Find the size of the largest class
    max_size = df['label'].value_counts().max()
    
    resampled = []
    for cls in df['label'].unique():
        cls_df = df[df['label'] == cls]
        if len(cls_df) < max_size:
            # Oversample with replacement
            cls_df = cls_df.sample(max_size, replace=True, random_state=random_state)
        resampled.append(cls_df)
    
    df_res = pd.concat(resampled, axis=0).sample(frac=1, random_state=random_state)  # shuffle
    
    X_res = df_res.drop(columns='label').values
    y_res = df_res['label'].values
    return X_res, y_res


# -------------------------------------------------
# Cross-validation
# -------------------------------------------------
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_scores = {'acc': [], 'bal': [], 'macro': [], 'weighted': []}
xgb_scores = {'acc': [], 'bal': [], 'macro': [], 'weighted': []}
ensemble_scores = {'acc': [], 'bal': [], 'macro': [], 'weighted': []}

fold = 1
for train_idx, val_idx in skf.split(X, y):
    print(f"\n----- Fold {fold} -----")
    
    X_tr, X_va = X[train_idx], X[val_idx]
    y_tr, y_va = y[train_idx], y[val_idx]
    
    # 1. Scale
    scaler = RobustScaler()
    X_tr = scaler.fit_transform(X_tr)
    X_va = scaler.transform(X_va)
    
    # 2. Manual oversampling (only on train)
    X_tr_res, y_tr_res = manual_oversample(X_tr, y_tr)
    print(f"  After oversampling → train samples: {X_tr_res.shape[0]} (was {X_tr.shape[0]})")
    
    # 3. Class weights (still useful)
    classes = np.unique(y_tr_res)
    cw = compute_class_weight('balanced', classes=classes, y=y_tr_res)
    cw_dict = dict(zip(classes, cw))
    sw = np.array([cw_dict[yy] for yy in y_tr_res])
    
    # 4. Train models
    rf = RandomForestClassifier(
        n_estimators=300,
        class_weight='balanced',
        n_jobs=-1,
        random_state=42
    )
    rf.fit(X_tr_res, y_tr_res)
    pred_rf = rf.predict(X_va)
    
    xgb_model = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective='multi:softprob',
        eval_metric='mlogloss',
        n_jobs=-1,
        random_state=42
    )
    xgb_model.fit(X_tr_res, y_tr_res, sample_weight=sw, verbose=False)
    pred_xgb = xgb_model.predict(X_va)
    
    # Soft Ensemble
    proba_rf  = rf.predict_proba(X_va)
    proba_xgb = xgb_model.predict_proba(X_va)
    proba_ens = (proba_rf + proba_xgb) / 2
    pred_ens  = np.argmax(proba_ens, axis=1)
    
    # 5. Metrics
    for name, pred, scores in [
        ('RF', pred_rf, rf_scores),
        ('XGB', pred_xgb, xgb_scores),
        ('Ensemble', pred_ens, ensemble_scores)
    ]:
        scores['acc'].append(accuracy_score(y_va, pred))
        scores['bal'].append(balanced_accuracy_score(y_va, pred))
        scores['macro'].append(f1_score(y_va, pred, average='macro'))
        scores['weighted'].append(f1_score(y_va, pred, average='weighted'))
    
    print(f"RF       → Acc: {rf_scores['acc'][-1]:.4f} | Bal: {rf_scores['bal'][-1]:.4f} | MacroF1: {rf_scores['macro'][-1]:.4f}")
    print(f"XGB      → Acc: {xgb_scores['acc'][-1]:.4f} | Bal: {xgb_scores['bal'][-1]:.4f} | MacroF1: {xgb_scores['macro'][-1]:.4f}")
    print(f"Ensemble → Acc: {ensemble_scores['acc'][-1]:.4f} | Bal: {ensemble_scores['bal'][-1]:.4f} | MacroF1: {ensemble_scores['macro'][-1]:.4f}")
    
    fold += 1

# ============================================================
# Final Results
# ============================================================
print("\n" + "="*65)
print("5-FOLD CV + MANUAL OVERSAMPLING  –  FINAL RESULTS")
print("="*65)

for name, scores in [('Random Forest', rf_scores), 
                     ('XGBoost', xgb_scores), 
                     ('Ensemble', ensemble_scores)]:
    print(f"\n{name}:")
    print(f"  Accuracy          : {np.mean(scores['acc']):.4f} ± {np.std(scores['acc']):.4f}")
    print(f"  Balanced Accuracy : {np.mean(scores['bal']):.4f} ± {np.std(scores['bal']):.4f}")
    print(f"  Macro F1          : {np.mean(scores['macro']):.4f} ± {np.std(scores['macro']):.4f}")
    print(f"  Weighted F1       : {np.mean(scores['weighted']):.4f} ± {np.std(scores['weighted']):.4f}")
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras import backend as K
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score, classification_report
from collections import defaultdict
import rasterio
from pathlib import Path
from tqdm.auto import tqdm

# ============================================================
# 0. Paths & Constants (adjust if needed)
# ============================================================
DATA_ROOT = Path("/home/23CSU413/Deep Learning/AgriFieldNet")
SOURCE_DIR = DATA_ROOT / "source"
TRAIN_LABEL_DIR = DATA_ROOT / "train_labels" / "train_labels"

BANDS = ["B01", "B02", "B03", "B04", "B05", "B06",
         "B07", "B08", "B8A", "B09", "B11", "B12"]

# Original → remapped label mapping (same as before)
ORIGINAL_LABELS = [1, 2, 3, 4, 5, 6, 8, 9, 13, 14, 15, 16, 36]
label_to_idx = {lab: idx for idx, lab in enumerate(ORIGINAL_LABELS)}
NUM_CLASSES = 13          # 0-12 after remapping
IGNORE_INDEX = 255        # we will use this for NoData (0)

# ============================================================
# 1. Helper functions to load one chip
# ============================================================
def load_chip(chip_id):
    folder = SOURCE_DIR / f"ref_agrifieldnet_competition_v1_source_{chip_id}"
    bands = []
    for b in BANDS:
        tif = folder / f"ref_agrifieldnet_competition_v1_source_{chip_id}_{b}_10m.tif"
        with rasterio.open(tif) as src:
            bands.append(src.read(1).astype(np.float32))
    image = np.stack(bands, axis=-1)          # (256, 256, 12)
    
    crop_path = TRAIN_LABEL_DIR / f"ref_agrifieldnet_competition_v1_labels_train_{chip_id}.tif"
    field_path = TRAIN_LABEL_DIR / f"ref_agrifieldnet_competition_v1_labels_train_{chip_id}_field_ids.tif"
    
    with rasterio.open(crop_path) as src:
        crop = src.read(1)
    with rasterio.open(field_path) as src:
        field = src.read(1)
    
    return image, crop, field


def remap_mask(crop):
    """Remap original crop labels to 0-12. NoData (0) → IGNORE_INDEX"""
    mask = np.full_like(crop, IGNORE_INDEX, dtype=np.uint8)
    for orig, idx in label_to_idx.items():
        mask[crop == orig] = idx
    return mask


# ============================================================
# 2. Build list of all train chip IDs
# ============================================================
def get_train_chip_ids():
    files = list(TRAIN_LABEL_DIR.glob("*_field_ids.tif"))
    ids = []
    for f in files:
        parts = f.stem.split("_")
        ids.append(parts[-3])
    return sorted(set(ids))

train_chip_ids = get_train_chip_ids()
print(f"Total train chips: {len(train_chip_ids)}")

# ============================================================
# 3. Create dataset of (image, mask) – only chips that have labels
# ============================================================
images = []
masks = []
field_ids_list = []          # keep field_id maps for later aggregation
chip_ids_kept = []

print("Loading chips (this takes a few minutes)...")
for cid in tqdm(train_chip_ids):
    try:
        image, crop, field = load_chip(cid)
        
        # Skip chips with almost no labeled pixels
        if np.sum(crop > 0) < 50:
            continue
            
        mask = remap_mask(crop)
        
        # Simple normalization per chip
        image = (image - image.mean(axis=(0,1), keepdims=True)) / (image.std(axis=(0,1), keepdims=True) + 1e-6)
        
        images.append(image)
        masks.append(mask)
        field_ids_list.append(field)
        chip_ids_kept.append(cid)
    except Exception as e:
        print(f"Skipping {cid}: {e}")

images = np.array(images, dtype=np.float32)
masks  = np.array(masks, dtype=np.uint8)

print(f"\nLoaded {len(images)} chips")
print("Images shape:", images.shape)
print("Masks shape :", masks.shape)

# ============================================================
# 4. Train / Validation split (by chip)
# ============================================================
idx = np.arange(len(images))
train_idx, val_idx = train_test_split(idx, test_size=0.2, random_state=42)

X_tr, X_va = images[train_idx], images[val_idx]
y_tr, y_va = masks[train_idx], masks[val_idx]
field_tr = [field_ids_list[i] for i in train_idx]
field_va = [field_ids_list[i] for i in val_idx]

print("Train chips:", len(X_tr), "| Val chips:", len(X_va))

# ============================================================
# 5. U-Net Model
# ============================================================
def conv_block(x, filters):
    x = layers.Conv2D(filters, 3, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(filters, 3, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    return x

def build_unet(input_shape=(256, 256, 12), num_classes=NUM_CLASSES):
    inputs = layers.Input(shape=input_shape)
    
    # Encoder
    c1 = conv_block(inputs, 32)
    p1 = layers.MaxPooling2D()(c1)
    
    c2 = conv_block(p1, 64)
    p2 = layers.MaxPooling2D()(c2)
    
    c3 = conv_block(p2, 128)
    p3 = layers.MaxPooling2D()(c3)
    
    c4 = conv_block(p3, 256)
    p4 = layers.MaxPooling2D()(c4)
    
    # Bottleneck
    bn = conv_block(p4, 512)
    
    # Decoder
    u4 = layers.UpSampling2D()(bn)
    u4 = layers.Concatenate()([u4, c4])
    c5 = conv_block(u4, 256)
    
    u3 = layers.UpSampling2D()(c5)
    u3 = layers.Concatenate()([u3, c3])
    c6 = conv_block(u3, 128)
    
    u2 = layers.UpSampling2D()(c6)
    u2 = layers.Concatenate()([u2, c2])
    c7 = conv_block(u2, 64)
    
    u1 = layers.UpSampling2D()(c7)
    u1 = layers.Concatenate()([u1, c1])
    c8 = conv_block(u1, 32)
    
    outputs = layers.Conv2D(num_classes, 1, activation='softmax')(c8)
    
    return models.Model(inputs, outputs, name="UNet")

unet = build_unet()
unet.summary()

# ============================================================
# 6. Custom loss that ignores NoData pixels
# ============================================================
def sparse_focal_loss(gamma=2.0, ignore_index=IGNORE_INDEX):
    def loss(y_true, y_pred):
        y_true = tf.cast(y_true, tf.int32)
        mask = tf.not_equal(y_true, ignore_index)
        
        y_true = tf.boolean_mask(y_true, mask)
        y_pred = tf.boolean_mask(y_pred, mask)
        
        y_true = tf.one_hot(y_true, depth=NUM_CLASSES)
        
        # Focal loss
        cross_entropy = -y_true * tf.math.log(tf.clip_by_value(y_pred, 1e-7, 1.0))
        weight = tf.pow(1 - y_pred, gamma)
        loss = weight * cross_entropy
        return tf.reduce_mean(tf.reduce_sum(loss, axis=-1))
    return loss

unet.compile(
    optimizer=optimizers.Adam(1e-4),
    loss=sparse_focal_loss(gamma=2.0),
    metrics=['accuracy']
)

# ============================================================
# 7. Train
# ============================================================
history = unet.fit(
    X_tr, y_tr,
    validation_data=(X_va, y_va),
    epochs=40,
    batch_size=4,                    # small because 256x256x12
    callbacks=[
        callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True, verbose=1),
        callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4, min_lr=1e-6, verbose=1)
    ],
    verbose=1
)

# ============================================================
# 8. Field-level aggregation & evaluation
# ============================================================
def predict_field_level(model, images, field_maps, true_masks):
    """
    For every field_id in the validation chips,
    average the predicted probabilities of its pixels.
    """
    field_true = {}
    field_proba = defaultdict(list)
    
    for img, field_map, true_mask in zip(images, field_maps, true_masks):
        pred = model.predict(img[np.newaxis, ...], verbose=0)[0]   # (256,256,13)
        
        unique_fids = np.unique(field_map)
        for fid in unique_fids:
            if fid == 0:
                continue
            pixels = (field_map == fid)
            if not np.any(pixels):
                continue
                
            # True label of this field (majority)
            true_labels = true_mask[pixels]
            true_labels = true_labels[true_labels != IGNORE_INDEX]
            if len(true_labels) == 0:
                continue
            true_lab = np.bincount(true_labels).argmax()
            
            # Average probability
            avg_proba = pred[pixels].mean(axis=0)
            
            field_true[fid] = true_lab
            field_proba[fid].append(avg_proba)
    
    # Average if a field appears in multiple chips
    y_true_list = []
    y_pred_list = []
    for fid, true_lab in field_true.items():
        avg_p = np.mean(field_proba[fid], axis=0)
        pred_lab = np.argmax(avg_p)
        y_true_list.append(true_lab)
        y_pred_list.append(pred_lab)
    
    return np.array(y_true_list), np.array(y_pred_list)


print("\nRunning field-level evaluation on validation chips...")
y_true_f, y_pred_f = predict_field_level(unet, X_va, field_va, y_va)

print("\n" + "="*60)
print("U-NET FIELD-LEVEL RESULTS")
print("="*60)
print(f"Number of fields evaluated: {len(y_true_f)}")
print(f"Accuracy          : {accuracy_score(y_true_f, y_pred_f):.4f}")
print(f"Balanced Accuracy : {balanced_accuracy_score(y_true_f, y_pred_f):.4f}")
print(f"Macro F1          : {f1_score(y_true_f, y_pred_f, average='macro'):.4f}")
print(f"Weighted F1       : {f1_score(y_true_f, y_pred_f, average='weighted'):.4f}")
print(classification_report(y_true_f, y_pred_f, zero_division=0))
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
from IPython.display import display

# ============================================================
# 1. Helper function to safely evaluate a model
# ============================================================
def evaluate_model(model, X, y, name):
    try:
        if hasattr(model, "predict_proba"):
            y_pred = model.predict(X)
        else:
            # Keras model
            y_pred = np.argmax(model.predict(X, verbose=0), axis=1)
        
        return {
            "Model": name,
            "Accuracy": accuracy_score(y, y_pred),
            "Balanced Accuracy": balanced_accuracy_score(y, y_pred),
            "Macro F1": f1_score(y, y_pred, average="macro"),
            "Weighted F1": f1_score(y, y_pred, average="weighted")
        }
    except Exception as e:
        print(f"Could not evaluate {name}: {e}")
        return None


# ============================================================
# 2. Try to find the correct validation sets
# ============================================================
# Field-level features (for RF, XGBoost, SVM, MLP...)
X_field = None
y_field = None

for x_name, y_name in [
    ("X_val", "y_val"),
    ("X_feat_va", "y_va"),
    ("X_va", "y_va"),
]:
    if x_name in globals() and y_name in globals():
        X_candidate = globals()[x_name]
        if len(X_candidate.shape) == 2:          # must be 2D
            X_field = X_candidate
            y_field = globals()[y_name]
            print(f"Using field-level validation set: {x_name}, {y_name} → {X_field.shape}")
            break

# Image patches (for CNNs)
X_img_val = None
y_img_val = None

for x_name, y_name in [
    ("X_img_va", "y_va"),
    ("X_va", "y_va"),
    ("X_img_val", "y_val"),
]:
    if x_name in globals() and y_name in globals():
        X_candidate = globals()[x_name]
        if len(X_candidate.shape) == 4:          # must be 4D (N,H,W,C)
            X_img_val = X_candidate
            y_img_val = globals()[y_name]
            # If y is one-hot, convert back
            if len(y_img_val.shape) > 1:
                y_img_val = np.argmax(y_img_val, axis=1)
            print(f"Using image validation set: {x_name}, {y_name} → {X_img_val.shape}")
            break


# ============================================================
# 3. List of models to try
# ============================================================
candidates = [
    # Classical
    ("rf_final", "Random Forest", "field"),
    ("rf", "Random Forest", "field"),
    ("xgb_final", "XGBoost", "field"),
    ("xgb_model", "XGBoost", "field"),
    ("xgb", "XGBoost", "field"),
    
    # CNNs
    ("model", "Stable CNN (32×32)", "image"),
    ("cnn", "Stable CNN (32×32)", "image"),
    
    # U-Net (special – already has field-level results)
    ("unet", "U-Net (Field-level)", "unet"),
]

results = []

print("\nEvaluating available models...\n")

for var_name, display_name, mode in candidates:
    if var_name not in globals():
        continue
    
    model = globals()[var_name]
    
    if mode == "field" and X_field is not None:
        res = evaluate_model(model, X_field, y_field, display_name)
        if res:
            results.append(res)
            print(f"✓ {display_name}")
            
    elif mode == "image" and X_img_val is not None:
        res = evaluate_model(model, X_img_val, y_img_val, display_name)
        if res:
            results.append(res)
            print(f"✓ {display_name}")
            
    elif mode == "unet":
        # U-Net was already evaluated at field level earlier
        # We just record the number we obtained
        results.append({
            "Model": "U-Net (Field-level)",
            "Accuracy": 0.385,
            "Balanced Accuracy": 0.108,
            "Macro F1": 0.098,
            "Weighted F1": 0.304
        })
        print(f"✓ U-Net (Field-level) – using previously computed field-level metrics")


# ============================================================
# 4. Create and display the table
# ============================================================
if not results:
    print("No models could be evaluated. Please check variable names.")
else:
    df = pd.DataFrame(results)
    
    # Remove duplicates (keep the first occurrence)
    df = df.drop_duplicates(subset=["Model"]).reset_index(drop=True)
    
    # Sort by Balanced Accuracy
    df = df.sort_values("Balanced Accuracy", ascending=False).reset_index(drop=True)
    
    print("\n" + "="*80)
    print("CROP CLASSIFICATION RESULTS (from currently loaded models)")
    print("="*80)
    
    display(df.style.format({
        "Accuracy": "{:.3f}",
        "Balanced Accuracy": "{:.3f}",
        "Macro F1": "{:.3f}",
        "Weighted F1": "{:.3f}"
    }).background_gradient(subset=["Balanced Accuracy", "Macro F1"], cmap="YlGn"))
    
    df.to_csv("live_results.csv", index=False)
    print("\nSaved to live_results.csv")
```
